```python
# Question one

word=input("Enter string : ")

total=0         # Initialize total to sum the digits in the string
text=""         # Initialize text to store the non-digit characters

# for loop loops on the string and checks every charecter 

for ch in word:
    if ch.isdigit():
        total+=int(ch)
    else:
        text+=ch
print("Sum of digits in the string is : ",total)
print("String is : ",text)

 
```

```python
# Question two

password=input("Enter password : ")
   
   #check if the password is between 8 and 16 in length, if not print the message and exit the program
if not 8 <= len(password) <= 16 :
    print ("Password must be between 8 and 16")
else:
    has_upper = False
    has_special = False
    special_characters = "!@#$%^&*()-+?_=,<>/"
    for ch in password:
        if ch.isupper():
            has_upper = True
        if ch in special_characters:
            has_special = True
    if has_upper and has_special:
        print("Password is strong")
    else:
        print("Password is weak")
```

```python
# Question five

students={"Ahmed":[85,90,88],"Ali":[93,87,95],"Sara":[95,90,85],"Fatima":[80,85,90],"Hassan":[75,80,85]}

grades = []

for student_scores in students.values():
    grades.append(student_scores) 
for score in grades:
        average = sum(score) / len(score)
        if average > 90:
            print("Name: ",list(students.keys())[grades.index(score)],"Average: ",average,"Honor Roll")
        elif average > 80:
            print("Name: ",list(students.keys())[grades.index(score)],"Average: ",average,)
        elif average > 70:
            print("Name: ",list(students.keys())[grades.index(score)],"Average: ",average,)
        else:
            print("Grade D")
```

```python
#question nine

number = int(input("Enter a number: "))
factorial = 1
sum1 = 0

for i in range(1, number + 1):
    factorial *= i
    sum1 += i

if factorial % sum1 == 0:
    print("Factorial:", factorial,"Sum:", sum1," ",factorial ,"is divisible by: ", sum1)
    print ("Result: ",factorial//sum1)
else:
    print("Factorial:", factorial,"Sum:", sum1," ",factorial ,"is not divisible by: ", sum1)
```

```python
#question ten

store = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
cart = []

def calculate_total(cart):
    total = 0
    for item in cart:
        total += store[item]
    return total

def add_to_cart(cart, item):
    if item in store:
        cart.append(item)
        print(f"Added '{item}' to the cart.")
    else:
        print(f"Item '{item}' is not available in the store.")

def remove_from_cart(cart, item):
    if item in cart:
        cart.remove(item)
        print(f"Removed '{item}' from the cart.")
    else:
        print(f"Item '{item}' is not in the cart.")

while True:
    action = input("Add, Remove, or Done? ").lower()

    if action == "done":
        break

    item = input("Enter item: ").lower()

    if action == "add":
        add_to_cart(cart, item)

    elif action == "remove":
        remove_from_cart(cart, item)

    else:
        print("Invalid action.")

print("Cart:", cart)
print("Total Bill: $", calculate_total(cart))
```

```python
#question eleven

word = input("Enter a word: ")

def is_palindrome(word):
    word = word.lower()  # Convert to lowercase for case-insensitive comparison
    word = word.replace(" ", "") 
    return word == word[::-1]  # Check if the word is the same forwards and backwards  
if is_palindrome(word):
    print(f"'{word}' is a palindrome.") 
else :
    print(f"'{word}' is not a palindrome.")
```

```python
#question thirteen

list1 = [1, 2, 7, 4, 5, 0]
peaks = []

for i in range(1, len(list1) - 1):
    if list1[i] > list1[i - 1] and list1[i] > list1[i + 1]:
        peaks.append(list1[i])

print("Peaks in the list are:", peaks)
```

```python
#question fourteen

d1 = {'A': 1, 'B': 2, 'C': 1}
temp = {}
for key, value in d1.items():
    if value not in temp:
        temp[value] = [key]
    else:
        temp[value].append(key)

print("Inverted dictionary:", temp)
```

```python
#question fifteen

text = input("Enter a string: ")
vowels = "aeiouAEIOU"

result = {}

for i in range(len(text)):
    if text[i] in vowels:
        if text[i] not in result:
            result[text[i]] = []

        result[text[i]].append(i)

print(result)
```
